# Calculate VPD for CNRM-CERFACS_CNRM-CM6-1-HR, MOHC_HadGEM3-GC31-MM, and NOAA-GFDL_GFDL-ESM4

Import packages

```python
from datetime import timedelta
import xarray as xr
import numpy as np
import pandas as pd
```

Calculate vpd for historical


```python
def calculate_vpd_his(model):
    his_file_path = '..\\VPD\\' + model + '\\historical\\historical_' + model + '_VPD.csv'
    all_his = pd.read_csv(his_file_path)
    # print(all_his)
    # all_his = pd.concat([ps_his, huss_his, tas_his], axis=1)
    all_his['Ta'] = all_his['tas'] - 273.15

    # calculate rh and vpd (Set RH higher than 100 to 100, and set VPD lower than 0 to 0)
    all_his['RH'] = 100 * (all_his['huss'] * all_his['ps'] / 1000) / (
                0.622 * (0.611 * np.exp(17.26938818 * (all_his['Ta'] / (237.3 + all_his['Ta'])))))
    all_his['RH'] = all_his['RH'].apply(lambda x: min(x, 100))

    all_his['VPD'] = (6.11 * np.exp(17.26938818 * (all_his['Ta'] / (237.3 + all_his['Ta'])))) * (
                1 - (all_his['RH'] / 100.0)).apply(lambda x: max(x, 0))
    
    print(f'filtered RH and VPD {model} historical\n {all_his}')
    
    save_path = '..\\VPD\\' + model + '\\historical\\historical_' + model + '_VPD_filtered.csv'
    all_his.to_csv(save_path,index='True')
```

Calculate vpd for ssp

```python
def calculate_vpd_ssp(model, ssp):
    
    ssp_file_path = '..\\VPD\\' + model + '\\' + ssp + '\\' + ssp + '_' + model + '_VPD.csv'
    all_ssp = pd.read_csv(ssp_file_path)
    # print(all_ssp)
    all_ssp['Ta'] = all_ssp['tas'] - 273.15

    # calculate rh and vpd (Set RH higher than 100 to 100, and set VPD lower than 0 to 0)
    all_ssp['RH'] = 100 * (all_ssp['huss'] * all_ssp['ps'] / 1000) / (
                0.622 * (0.611 * np.exp(17.26938818 * (all_ssp['Ta'] / (237.3 + all_ssp['Ta'])))))
    all_ssp['RH'] = all_ssp['RH'].apply(lambda x: min(x, 100))
    all_ssp['VPD'] = (6.11 * np.exp(17.26938818 * (all_ssp['Ta'] / (237.3 + all_ssp['Ta'])))) * (
                1 - (all_ssp['RH'] / 100.0)).apply(lambda x: max(x, 0))
    
    print(f'filtered RH and VPD {model} {ssp}\n {all_ssp}')
    
    save_path = '..\\VPD\\' + model + '\\' + ssp + '\\' + ssp + '_' + model + '_VPD_filtered.csv'
    all_ssp.to_csv(save_path,index='True')
```

Calculate and save VPD for all three models

```python
model = 'CNRM-CERFACS_CNRM-CM6-1-HR'
calculate_vpd_his(model)
ssp = 'ssp126'
calculate_vpd_ssp(model, ssp)
ssp = 'ssp585'
calculate_vpd_ssp(model, ssp)

model = 'MOHC_HadGEM3-GC31-MM'
calculate_vpd_his(model)
ssp = 'ssp126'
calculate_vpd_ssp(model, ssp)
ssp = 'ssp585'
calculate_vpd_ssp(model, ssp)

model = 'NOAA-GFDL_GFDL-ESM4'
calculate_vpd_his(model)
ssp = 'ssp126'
calculate_vpd_ssp(model, ssp)
ssp = 'ssp585'
calculate_vpd_ssp(model, ssp)
ssp = 'ssp245' # For 'NOAA-GFDL_GFDL-ESM4'
calculate_vpd_ssp(model, ssp)
ssp = 'ssp370' # For 'NOAA-GFDL_GFDL-ESM4'
calculate_vpd_ssp(model, ssp)
```
